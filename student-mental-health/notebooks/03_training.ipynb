### Your code

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
)

probabilities_rf = rf.predict_proba(X_test)
probabilities_lr = lr.predict_proba(X_test)
probabilities_gb = gb.predict_proba(X_test)

resultats = pd.DataFrame({
    'Model': [
        'Random Forest',
        'Logistic Regression',
        'Gradient Boosting'
    ],
    'Accuracy': [
        accuracy_score(y_test, predictions_rf),
        accuracy_score(y_test, predictions_lr),
        accuracy_score(y_test, predictions_gb)
    ],
    'F1-Score': [
        f1_score(y_test, predictions_rf, average="macro"),
        f1_score(y_test, predictions_lr, average="macro"),
        f1_score(y_test, predictions_gb, average="macro")
    ],
    'Recall': [
        recall_score(y_test, predictions_rf, average="macro"),
        recall_score(y_test, predictions_lr, average="macro"),
        recall_score(y_test, predictions_gb, average="macro")
    ],
    'Precision': [
        precision_score(y_test, predictions_rf, average="macro"),
        precision_score(y_test, predictions_lr, average="macro"),
        precision_score(y_test, predictions_gb, average="macro")
    ],
    'ROC AUC': [
    roc_auc_score(y_test, probabilities_rf, average="macro", multi_class="ovr", labels=rf.classes_),
    roc_auc_score(y_test, probabilities_lr, average="macro", multi_class="ovr", labels=lr.classes_),
    roc_auc_score(y_test, probabilities_gb, average="macro", multi_class="ovr", labels=gb.classes_)
]
})

display(resultats)

```

## 6. Compare performance and stability

### What to do
- Make a single comparison table and plot for each model.
- Compare average validation scores, spread, training time, and interpretability.
- Note whether differences are small or consistent.

### Why it matters
A slightly higher score may not justify a slower, more complex, or less interpretable model.

### Example of expected output

**Illustrative conclusion:** “Random Forest has the highest mean macro-F1 but only a small margin; both it and Logistic Regression will be tuned.”

*The sample above is illustrative, not a result from your dataset.*

### Success checklist
- [ ] Side-by-side comparison produced.
- [ ] At least one figure made.
- [ ] Two promising candidates selected.


### Your code
Write your Python code in the empty cell below. No solution is supplied.

```python
from sklearn.model_selection import cross_validate
import pandas as pd

models = {
    "Random Forest": rf,
    "Logistic Regression": lr,
    "Gradient Boosting": gb
}

results = []

for name, model in models.items():
    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=5,
        scoring="f1_macro",
        return_train_score=True
    )

    results.append({
        "Model": name,
        "f1_score": scores["test_score"].mean(),
    })

comparison = pd.DataFrame(results)
display(comparison.sort_values("f1_score", ascending=False))
```
