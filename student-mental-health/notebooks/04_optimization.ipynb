### Your code

```python
import joblib
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
)
from sklearn.model_selection import GridSearchCV
import os

DATA_PATH = Path("/home/mestry/Documents/youcode/Pr-diction-du-risque-de-sant-mentale-chez-les-tudiants/student-mental-health/data/processed")
MODEL_PATH = "/home/mestry/Documents/youcode/Pr-diction-du-risque-de-sant-mentale-chez-les-tudiants/student-mental-health/models/student_mental_health_pipeline.joblib"

X_train = pd.read_csv(DATA_PATH / "X_train.csv")
X_test = pd.read_csv(DATA_PATH / "X_test.csv")
y_train = pd.read_csv(DATA_PATH / "y_train.csv").squeeze("columns")
y_test = pd.read_csv(DATA_PATH / "y_test.csv").squeeze("columns")

scoring = "f1_macro"

print(f"Training rows: {len(X_train)} | Test rows: {len(X_test)}")
```

**My answer:** The most important trade-off is good performance and simplicity. I want a model that predicts well and is easy to explain because this is an educational project.

## 2. Design hyperparameter search spaces

### What to do
- Choose meaningful hyperparameters for each selected model.
- Define practical ranges and avoid unnecessarily expensive searches.
- Consider GridSearchCV for small grids or RandomizedSearchCV for wider spaces.

### Why it matters
Hyperparameters control model behavior; structured search helps find better settings with controlled resources.

### Example of expected output

**Illustrative search description:** `Random Forest: number of trees, maximum depth, minimum samples per leaf`

*The sample above is illustrative, not a result from your dataset.*

### Success checklist
- [ ] Search space defined.
- [ ] Search method justified.


### Your code
Write your Python code in the empty cell below. No solution is supplied.

```python
lr = GridSearchCV(
    LogisticRegression(max_iter=2000, random_state=42),
    {
        "C": [0.1, 1, 10],
        "class_weight": [None, "balanced"]
    },
    cv=5,
    scoring="f1"
)

rf = GridSearchCV(
    RandomForestClassifier(random_state=42),
    {
        "n_estimators": [100, 200],
        "max_depth": [None, 10],
        "min_samples_leaf": [1, 2]
    },
    cv=5,
    scoring="f1"
)


```

**My answer:** Searching too many combinations takes a long time and can overfit the training data. A small, meaningful search is easier to understand and is enough for this project.

## 3. Run tuning using training-only CV

### What to do
- Run hyperparameter optimization on the 80% training partition.
- Make sure preprocessing and any resampling are **inside** the CV pipeline.
- Record best parameters, best mean CV score, and execution time.

### Why it matters
Hyperparameter tuning is a form of model selection: it must never learn from the final test set.

### Example of expected output

**Illustrative:** `Best CV macro-F1: 0.76 | Best parameters: {...}`

*The sample above is illustrative, not a result from your dataset.*

### Success checklist
- [ ] Search completes.
- [ ] Best CV result captured.
- [ ] Test set remains untouched.

### Watch out
Only use the test set once after you have chosen and finalized the model. If you repeatedly tune after seeing test results, it is no longer an unbiased final test.


### Your code
Write your Python code in the empty cell below. No solution is supplied.

```python


lr = GridSearchCV(
    LogisticRegression(max_iter=2000, random_state=42),
    {
        "C": [0.01,0.1, 0.5,1,2,3,4,5, 10, 100],
        "class_weight": [None, "balanced"]
    },
    cv=5,
    scoring="f1_macro"
)

lr.fit(X_train, y_train)

print("Logistic Regression")
print("Best Parameters:", lr.best_params_)
print("Best CV Score:", lr.best_score_)


rf = GridSearchCV(
    RandomForestClassifier(random_state=42),
    {
        "n_estimators": [100, 200],
        "max_depth": [None, 10],
        "min_samples_leaf": [1, 2]
    },
    cv=5,
    scoring="f1_macro"
)

rf.fit(X_train, y_train)

print("Random Forest")
print("Best Parameters:", rf.best_params_)
print("Best CV Score:", rf.best_score_)
```

**My answer:** Tuning is useful when the tuned macro-F1 is higher than the baseline score. If the improvement is small, the baseline was already good. I compare the scores using cross-validation, not the test set.

## 5. Evaluate the final model ONCE on test data

### What to do
- Apply the finalized pipeline to untouched 20% test rows.
- Calculate appropriate accuracy, precision, recall, and F1.
- Add ROC-AUC only if suitable.
- Produce a clearly labeled confusion matrix.
- Compare final test performance with CV results.

### Why it matters
This is the closest estimate you have of performance on new records from a similar source.

### Example of expected output

**Illustrative:** `Test accuracy: 0.79 | Test macro-F1: 0.72`

`Confusion matrix: actual vs predicted labels`

*The sample above is illustrative, not a result from your dataset.*

### Success checklist
- [ ] Only final chosen pipeline tested.
- [ ] Test metrics and confusion matrix recorded.
- [ ] CV-versus-test difference explained.

### Watch out
An educational prototype with high test accuracy is still not a validated clinical screening or diagnosis tool.


### Your code
Write your Python code in the empty cell below. No solution is supplied.

```python


best_model = lr.best_estimator_
best_name = "Logistic Regression"
cv_score = lr.best_score_

y_pred = best_model.predict(X_test)

print("Best Model:", best_name)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, average="macro"))
print("Recall:", recall_score(y_test, y_pred, average="macro"))
print("F1:", f1_score(y_test, y_pred, average="macro"))

cm = confusion_matrix(y_test, y_pred, labels=best_model.classes_)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=best_model.classes_,
    yticklabels=best_model.classes_
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()


print("CV F1:", cv_score)
print("Test F1:", f1_score(y_test, y_pred, average="macro", zero_division=0))
```

**My answer:** The model generalizes reasonably if its test score is close to its cross-validation score. Some uncertainty remains because the dataset is limited and may not represent all students.
